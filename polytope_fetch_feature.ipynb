# Feature extraction with polytope
Example for how to extract model level data from polytope using feature extraction. To use polytope put your ECMWF web API key, https://api.ecmwf.int/v1/key/ in a ~/.polytopeapirc file as 

{
    "user_key"   : "some_hash",
    "user_email" : "foo.bar@somewhere.ok"
}


```python
# Import necessary libraries
import earthkit.data
from polytope.api import Client
from datetime import datetime
from versions import UrbanAirData
```

```python
# Get the latest version
uad = UrbanAirData().urls["9.0"]
```

```python
# Define the request
dt = uad["metadata"]["date"]
dt = datetime.strptime(dt, "%Y-%m-%dT%H:%M:%SZ")
request= {
    "class": "d1",
    "dataset": "on-demand-extremes-dt",
    "expver": uad["metadata"]["fdb"]["expver"],
    "stream": "oper",
    "date": dt.strftime("%Y%m%d"),
    "time": dt.strftime("%H%M"),
    "type": "fc",
    "georef": uad["metadata"]["fdb"]["georef"],
    "levtype": "ml",
    "step": 0,
    "levelist": "90",
    "param": 129,
    "timespan": "none",
      "feature": {
               "type": "polygon",
               "shape": [
                   [
                       48.0,
                       1.0
                   ],
                   [
                       48.0,
                       3.0
                   ],
                   [
                       49.0,
                       3.0
                   ],
                   [
                       49.0,
                       1.0
                   ]
               ],
      "axes": [
                   "latitude",
                   "longitude"
               ]
    }
}
```

```python
client = Client(address=uad["metadata"]["polytope"]["url"])
data = earthkit.data.from_source("polytope",
                                 uad["metadata"]["polytope"]["collection"],
                                 request, address=uad["metadata"]["polytope"]["url"],
                                 stream=False)
```
